```python
import pandas as pd
import numpy as np

files = {
    2024: "/Users/valeriia/Desktop/Ala-Tash DA/Строительные материалы 2024.xlsm",
    2025: "/Users/valeriia/Desktop/Ala-Tash DA/Строительные материалы 2025.xlsm",
    2026: "/Users/valeriia/Desktop/Ala-Tash DA/Строительные материалы 2026.xlsm"
}

order_columns = [
    "№ заказа",
    "Анкета",
    "Ф.И.О.",
    "Телефоны",
    "Установщик",
    "Принят",
    "Оговорено",
    "Отпущен",
    "Сумма, сом",
    "Скидка, сом",
    "Со скидкой",
    "Авансы",
    "Остаток",
    "Долг"
]

orders_list = []
items_list = []

for year, file_path in files.items():

    df = pd.read_excel(
        file_path,
        sheet_name="Статистика",
        engine="openpyxl"
    )

    # Очищаем переносы строк и лишние пробелы в названиях
    df.columns = (
        df.columns
        .astype(str)
        .str.replace("\n", " ", regex=False)
        .str.replace("\r", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    # Оставляем только строки, где есть номер заказа
    df = df[df["№ заказа"].notna()].copy()

    # Номер заказа сохраняем как текст
    df["№ заказа"] = (
        df["№ заказа"]
        .astype(str)
        .str.strip()
    )

    df["Год"] = year

    # Уникальный ключ: год + номер заказа
    df["Order_ID"] = (
        df["Год"].astype(str)
        + "_"
        + df["№ заказа"]
    )

    # Таблица заказов
    orders_year = df[order_columns + ["Год", "Order_ID"]].copy()

    orders_list.append(orders_year)

     # В файле предусмотрено до 10 товарных позиций
    for position in range(10):

        # Для первой позиции суффикса нет
        suffix = "" if position == 0 else f".{position}"

        name_col = f"Наименование{suffix}"
        unit_col = f"Ед. измер.{suffix}" if position > 0 else "Ед. измер."
        qty_col = f"Кол-во{suffix}"
        price_col = f"Цена, сом{suffix}"
        discounted_price_col = (
            f"Цена со скидкой, сом{suffix}"
        )

        # У столбцов суммы особая нумерация:
        # первая товарная сумма называется "Сумма, сом.1"
        sum_number = position + 1
        sum_col = f"Сумма, сом.{sum_number}"

        discounted_sum_col = (
            "Сумма со скидкой, сом"
            if position == 0
            else f"Сумма со скидкой, сом.{position}"
        )

        required_columns = [
            name_col,
            unit_col,
            qty_col,
            price_col,
            discounted_price_col,
            sum_col,
            discounted_sum_col
        ]

        # Проверяем наличие блока в конкретном файле
        if not all(col in df.columns for col in required_columns):
            print(
                f"{year}: пропущен блок позиции {position + 1}, "
                "не все столбцы найдены"
            )
            continue

        item_block = df[
            ["Order_ID", "Год", "№ заказа"] + required_columns
        ].copy()

        item_block.columns = [
            "Order_ID",
            "Год",
            "№ заказа",
            "Наименование",
            "Ед. измер.",
            "Кол-во",
            "Цена, сом",
            "Цена со скидкой, сом",
            "Сумма позиции, сом",
            "Сумма позиции со скидкой, сом"
        ]

        item_block["Позиция"] = position + 1

        # Оставляем строки, где действительно указано наименование
        item_block = item_block[
            item_block["Наименование"].notna()
        ].copy()

        items_list.append(item_block)

        sales_orders = pd.concat(
    orders_list,
    ignore_index=True
)

sales_items = pd.concat(
    items_list,
    ignore_index=True
)

print("Таблица заказов:", sales_orders.shape)
print("Таблица товарных позиций:", sales_items.shape)

print(
    sales_orders.groupby("Год")
    .size()
)
print(
    sales_items.groupby("Год")
    .size()
)

sales_items.head(20)

sales_items[
    sales_items["Order_ID"] == "2024_0,1"
]

order_numeric_columns = [
    "Сумма, сом",
    "Скидка, сом",
    "Со скидкой",
    "Авансы",
    "Остаток",
    "Долг"
]

for col in order_numeric_columns:
    sales_orders[col] = pd.to_numeric(
        sales_orders[col],
        errors="coerce"
    )

item_numeric_columns = [
    "Кол-во",
    "Цена, сом",
    "Цена со скидкой, сом",
    "Сумма позиции, сом",
    "Сумма позиции со скидкой, сом"
]

for col in item_numeric_columns:
    sales_items[col] = pd.to_numeric(
        sales_items[col],
        errors="coerce"
    )

date_columns = [
    "Принят",
    "Оговорено",
    "Отпущен"
]

for col in date_columns:
    sales_orders[col] = pd.to_datetime(
        sales_orders[col],
        errors="coerce"
    )
sales_orders.to_excel(
    "/Users/valeriia/Desktop/Ala-Tash DA/Sales_Orders_Master.xlsx",
    index=False
)

sales_items.to_excel(
    "/Users/valeriia/Desktop/Ala-Tash DA/Sales_Items_Master.xlsx",
    index=False
)
```

```python
import pandas as pd
import numpy as np

sales_clean = sales.copy()
mask = (
    (sales_clean["Принят"].dt.year < 2024) |
    (sales_clean["Принят"].dt.year > 2026)
)

sales_clean.loc[mask, "Принят"] = pd.NaT
sales_clean["Год_даты"] = sales_clean["Принят"].dt.year
sales_clean["Месяц_номер"] = sales_clean["Принят"].dt.month
months = {
    1:"Январь",
    2:"Февраль",
    3:"Март",
    4:"Апрель",
    5:"Май",
    6:"Июнь",
    7:"Июль",
    8:"Август",
    9:"Сентябрь",
    10:"Октябрь",
    11:"Ноябрь",
    12:"Декабрь"
}

sales_clean["Месяц"] = sales_clean["Месяц_номер"].map(months)
sales_clean["Квартал"] = sales_clean["Принят"].dt.quarter
sales_clean["Статус оплаты"] = np.select(
    [
        sales_clean["Остаток"] > 0,
        sales_clean["Остаток"] == 0,
        sales_clean["Остаток"] < 0
    ],
    [
        "Есть долг",
        "Оплачен",
        "Переплата"
    ],
    default="Нет данных"
)
sales_clean["Есть скидка"] = np.where(
    sales_clean["Скидка, сом"] > 0,
    "Да",
    "Нет"
)
sales_clean["Категория чека"] = pd.cut(
    sales_clean["Сумма, сом"],
    bins=[0,10000,30000,50000,100000,500000,10000000],
    labels=[
        "до 10 тыс.",
        "10–30 тыс.",
        "30–50 тыс.",
        "50–100 тыс.",
        "100–500 тыс.",
        "500 тыс.+"
    ]
)
sales_clean["Дней выполнения"] = (
    sales_clean["Отпущен"] -
    sales_clean["Принят"]
).dt.days
sales_clean.info()
sales_clean.head()
sales_clean.to_excel(
    "/Users/valeriia/Desktop/Ala-Tash DA/sales_orders_clean.xlsx",
    index=False
)
```
