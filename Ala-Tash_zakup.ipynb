```python
import pandas as pd

# ---------- Авто ----------
auto = pd.read_excel(
    "/Users/valeriia/Desktop/Ala-Tash DA/Закуп (1).xlsx",
    sheet_name="Авто"
)

# Удаляем служебные строки
auto = auto[auto["Дата поставки"].notna()].copy()

auto = auto.rename(columns={
    "Название": "Материал",
    "Кол-во": "Кол-во, кв.м."
})

auto["Способ доставки"] = "Авто"
auto["Откуда"] = None
```

```python
import os

os.listdir("/Users/valeriia/Desktop/Ala-Tash DA")
```

```python
rail = pd.read_excel(
    "/Users/valeriia/Desktop/Ala-Tash DA/Закуп (1).xlsx",
    sheet_name="Ж.Д."
)

rail = rail[rail["Приход на станцию"].notna()].copy()

rail = rail.rename(columns={
    "Приход на станцию": "Дата поставки"
})

rail["Размеры"] = None
rail["Фактура"] = None
rail["Способ доставки"] = "ЖД"
```

```python
auto = auto[
    [
        "Дата поставки",
        "Способ доставки",
        "Материал",
        "Изделие",
        "Размеры",
        "Фактура",
        "Кол-во, кв.м.",
        "Откуда"
    ]
]

rail = rail[
    [
        "Дата поставки",
        "Способ доставки",
        "Материал",
        "Изделие",
        "Размеры",
        "Фактура",
        "Кол-во, кв.м.",
        "Откуда"
    ]
]
```

```python
purchases = pd.concat(
    [auto, rail],
    ignore_index=True
)
```
